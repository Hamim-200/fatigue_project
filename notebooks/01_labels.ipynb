```python
import os
import pandas as pd

base = r"N:\CSE498R\fatigue_project\data\Data"

def load_labels(user):
    df = pd.read_csv(os.path.join(base, user, "usercondition.tsv"), sep="\t", index_col=False)
    df = df.loc[:, ~df.columns.str.startswith("Unnamed")]
    df["Time"] = pd.to_datetime(df["Time"])
    df["user"] = user
    return df

labels = pd.concat([load_labels("user1"), load_labels("user2")], ignore_index=True)
print(labels.shape)
labels.head(10)
```

```python
# Count each label and check session length

for col in ["Fatigue_Val", "PAM_Val", "Stress_Val", "Energy_Val", "Pleasant_Val"]:
    print(col)
    print(labels.groupby("user")[col].value_counts())
    print()

print(labels.groupby("user")["Time"].agg(["min", "max", "count"]))
print(labels.groupby("user")["Time"].diff().describe())
```

```python
# Compare fatigue with energy

pd.crosstab(labels["Fatigue_Val"], labels["Energy_Val"])
```
